```python
Input:
    Online item sequence IPS
    Single container (bin)

Output:
    Packing solution PS or null

-------------------------------------------------
1. Initialization
-------------------------------------------------
1.1  PS ← ∅

-------------------------------------------------
2. Main loop
-------------------------------------------------
2.1  while items are arriving or IPS ≠ null do
2.2      Update IPS
2.3      itemplaced ← 0
2.4      P ← ∅          // candidate placements

-------------------------------------------------
3. Extract corner points
-------------------------------------------------
3.1      Compute heightmap h(x,y) from container state
3.2      CP ← {(x, y, h(x,y)) for all (x,y)}

3.3      Sort CP in increasing order of (z, x, y)

-------------------------------------------------
4. Generate candidate placements
-------------------------------------------------
4.1      for each corner point p ∈ CP do
4.2          for each visible item i ∈ IPS (up to Kb) do
4.3              for each allowed rotation r of item i do
4.4                  if item i with rotation r fits at p
                         and does not overlap any placed item
4.5                  then
4.6                      evaluate placement score
4.7                      add (i, p, r) to P

-------------------------------------------------
5. Select placement
-------------------------------------------------
5.1      if P ≠ ∅ then
5.2          select best placement from P
5.3          place item
5.4          update container state
5.5          remove item from IPS
5.6          itemplaced ← 1

-------------------------------------------------
6. Failure handling
-------------------------------------------------
6.1      if itemplaced = 0 then
6.2          return null

-------------------------------------------------
7. End while
-------------------------------------------------
8. return PS
```

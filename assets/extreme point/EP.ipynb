```python
Input:
    - Item set I = {1, ..., n}
    - Container dimensions (L, W, H)
    - Set of allowed rotations R

Output:
    - Packing solution PS or null

-------------------------------------------------
1. Initialize:
-------------------------------------------------
1.1  EP ← {(0, 0, 0)}          // initial extreme point at container origin
1.2  PS ← ∅                   // packing solution
1.3  U ← I                    // unpacked items

-------------------------------------------------
2. Main loop:
-------------------------------------------------
2.1  while U ≠ ∅ do
2.2      P ← ∅                // candidate placements

-------------------------------------------------
3. Generate candidate placements:
-------------------------------------------------
3.1      for each item i ∈ U do
3.2          for each extreme point p ∈ EP do
3.3              for each rotation r ∈ R do
3.4                  if item i with rotation r fits at point p
                         and does not overlap any placed item
                         and lies inside the container
3.5                  then
3.6                      compute placement score (e.g., fill ratio)
3.7                      add (i, p, r, score) to P

-------------------------------------------------
4. Select placement:
-------------------------------------------------
4.1      if P = ∅ then
4.2          return null      // no feasible placement
4.3      select best placement (i*, p*, r*) from P
         according to placement rule

-------------------------------------------------
5. Place item and update EP set:
-------------------------------------------------
5.1      place item i* at point p* with rotation r*
5.2      add (i*, p*, r*) to PS
5.3      remove i* from U

-------------------------------------------------
6. Update extreme points:
-------------------------------------------------
6.1      remove from EP all points dominated by the placed item
6.2      generate new extreme points induced by item i*:
             - (x_i + dx_i, y_i, z_i)
             - (x_i, y_i + dy_i, z_i)
             - (x_i, y_i, z_i + dz_i)
6.3      remove invalid extreme points:
             - outside container
             - inside an item
             - dominated by another EP

-------------------------------------------------
7. End while
-------------------------------------------------
8. return PS
```

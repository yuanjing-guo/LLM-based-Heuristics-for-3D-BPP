```python
Input:
    I = {i1, i2, ..., in}          // set of items (boxes)
    B                              // 3D bin (container)
    C(B)                           // capacity constraints of the bin

Output:
    P                              // placement layout
    U                              // set of unplaced items

Initialize:
    P ← ∅
    U ← ∅

for each item i ∈ I do
    Generate candidate positions CP(i, B)

    placed ← false

    for each position p ∈ CP(i, B) do
        if Feasible(i, p, P, B) then
            Place item i at position p
            Update placement layout P
            placed ← true
            break                  // First-Fit decision
        end if
    end for

    if placed = false then
        U ← U ∪ {i}                // item cannot be placed by FF
    end if
end for

Apply Wall-Building heuristic to U and B
Update placement layout P

return P, U
```

```python
best fit is a optimized version of first fit.
Input:
    I = {i1, i2, ..., in}          // set of items
    B                              // 3D bin
    C(B)                           // bin constraints

Output:
    P                              // placement layout
    U                              // unplaced items

Initialize:
    P ← ∅
    U ← ∅

for each item i ∈ I do
    Generate candidate positions CP(i, B)

    best_position ← null
    best_score ← -∞

    for each position p ∈ CP(i, B) do
        if Feasible(i, p, P, B) then
            score ← SpaceScore(i, p, P, B)
            if score > best_score then
                best_score ← score
                best_position ← p
            end if
        end if
    end for

    if best_position ≠ null then
        Place item i at best_position
        Update placement layout P
    else
        U ← U ∪ {i}
    end if
end for

Apply Wall-Building heuristic to U and B
Update placement layout P

return P, U
```
